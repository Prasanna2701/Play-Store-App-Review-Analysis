#**Cleaning the Data Frame 1**#
**CONVERT TO NUMERAL DATA**





```python
# Retrieve the unique values from the 'Reviews' column
df1['Reviews'].unique()
```

```python
# Convert the 'Reviews' column to numeric values, coercing any non-numeric values to NaN
df1['Reviews'] = pd.to_numeric(df1['Reviews'],errors='coerce')
```

**Fix the Size**

```python
# Get the unique values in the 'Size' column
df1['Size'].unique()
```

```python
def fix_size(s):
  try:
    if s.endswith('M'):
      # If the string ends with 'M', convert the value directly to float
      return float(s[:-1])
    elif s.endswith('k'):
      # If the string ends with 'k', convert it to megabytes by dividing by 1024
      return (float(s[:-1]) / 1024)
    elif s == '1,000+':
      # For the special case '1,000+', return 1.0
      return 1.0
    else:
      # Return NaN for any other cases
      return np.nan
  except:
     # Return NaN if an exception occurs during conversion
      return np.nan
```

**Fix the  all elements in Size columns**

```python
# Apply the 'fix_size' function to each element in the 'Size' column
df1['Size']= df1['Size'].apply(fix_size)
```

```python
# Get the unique values in the 'Size' column
df1['Size'].unique()
```

```python
# Display basic information including column names, non-null counts, and data types of each column.
df1.info()
```

```python
# Retrieve all unique values from the 'Price' column
df1['Price'].unique()
```

```python
def fix_Price(A):
  try:
    if A.startswith('$'):
        # Convert the string to a float, excluding the '$' sign
      return float(A[1:])
    else:
      # Return 0.0 if the string does not start with '$'
      return 0.0
  except:
     # Return NaN if there is an error during conversion
    return np.nan
```

```python
# Display basic information including column names, non-null counts, and data types of each column.
df1.info()
```

```python
# Apply the 'fix_Price' function to each value in the 'Price' column
df1['Price']=df1['Price'].apply(fix_Price)
```

```python
# Display basic information including column names, non-null counts, and data types of each column.
df1.info()
```
